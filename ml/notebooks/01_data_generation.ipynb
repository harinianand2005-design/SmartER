# SmartER Synthetic ER Data Generation

This notebook creates reproducible synthetic hourly emergency-room operations data. It contains no real patient information and is intended only for academic demonstration.

```python
from pathlib import Path
import sys

root = Path.cwd()
while not (root / "dataset").exists():
    root = root.parent
sys.path.insert(0, str(root))

from ml.data.generator import generate_dataset

output_path = root / "dataset/generated/synthetic_er_hourly.csv"
output_path.parent.mkdir(parents=True, exist_ok=True)
data = generate_dataset(periods=3 * 365 * 24, seed=42)
data.to_csv(output_path, index=False)
print(data.shape)
data.head()
```

The fixed seed makes the generated source dataset reproducible. Synthetic relationships are assumptions for pipeline development and do not represent real hospital statistics.
