# SmartER Feature Engineering and Chronological Splits

Features are generated from historical values only. Future arrival columns are labels for later ML phases and are excluded from the fitted preprocessing inputs.

```python
from pathlib import Path
import sys

root = Path.cwd()
while not (root / "dataset").exists():
    root = root.parent
sys.path.insert(0, str(root))

import pandas as pd
from ml.preprocessing.features import engineer_features
from ml.preprocessing.pipeline import DataPreprocessor, chronological_split

raw = pd.read_csv(root / "dataset/generated/synthetic_er_hourly.csv")
featured = engineer_features(raw).dropna(subset=["next_hour_arrivals", "next_3hr_arrivals", "next_6hr_arrivals", "next_24hr_arrivals"])
splits = chronological_split(featured)
preprocessor = DataPreprocessor().fit(splits["training"])
processed = {name: preprocessor.transform(frame) for name, frame in splits.items()}
{name: frame.shape for name, frame in processed.items()}
```

The preprocessor learns medians and categorical levels from the training split only. Validation and test rows remain chronological and are transformed with those training-derived values.
